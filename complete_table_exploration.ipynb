## 📋 TASK 2: Load Selected Tables into SQL

**Your Task:**
1. Create SQL database
2. Load ONLY the 3 relevant tables
3. Create indexes for optimization
4. Document the decision

````python
# Create database connection
conn = sqlite3.connect('data/processed/vendor_analysis.db')
print("✓ Database connection established")
````

````python
# Load ONLY the 3 selected tables
print("Loading selected tables into SQL...\n")

purchases.to_sql('purchases', conn, if_exists='replace', index=False)
print(f"✓ purchases: {len(purchases):,} rows")

sales.to_sql('sales', conn, if_exists='replace', index=False)
print(f"✓ sales: {len(sales):,} rows")

vendor_invoice.to_sql('vendor_invoice', conn, if_exists='replace', index=False)
print(f"✓ vendor_invoice: {len(vendor_invoice):,} rows")

print("\n✓ All selected tables loaded into SQL database")
````

````python
# Create indexes for query optimization
print("Creating indexes...")

conn.execute("CREATE INDEX IF NOT EXISTS idx_purchases_inventory ON purchases(InventoryId)")
conn.execute("CREATE INDEX IF NOT EXISTS idx_purchases_vendor ON purchases(VendorNumber)")
conn.execute("CREATE INDEX IF NOT EXISTS idx_sales_inventory ON sales(InventoryId)")
conn.execute("CREATE INDEX IF NOT EXISTS idx_invoice_vendor ON vendor_invoice(VendorNumber)")
conn.commit()

print("✓ Indexes created for fast queries")
````

````python
# Clear from memory - we'll query from SQL now
del purchases, sales, vendor_invoice, begin_inv, end_inv, purchase_prices
print("✓ DataFrames cleared from memory")
print("\nFrom now on, we query data from SQL database")
````
